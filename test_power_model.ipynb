```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
percentage = np.arange(0, 1, 0.05)
```

```python
percentage
```

```python
def dBm2w(dBm_value):
    # convert dBm to milliwatts
    # milliwatts to Watts
    p = 10 ** (dBm_value / 10)
    return p * 1e-3
```

```python
p_static = 51
p_max_DU = 53
# delta_DU = 0.44
delta_DU = 0.2
```

```python
def func_power(percentage):
#     tmp = math.sqrt(percentage*2)
    tmp = math.tanh(percentage*3)
#     tmp = percentage ** 2
    return dBm2w(p_static + delta_DU * p_max_DU * tmp)
```

```python
max_p = func_power(1)
power_list = [func_power(item) / max_p for item in percentage]
```

```python
plt.plot(percentage, power_list)
print(max_p)
```
